## 2.1 Dasar Teori
Beberapa dasar teori yang meliputi praktikum ini adalah:

1. Pengertian Perulangan `for`

   Perulangan `for` pada Python adalah struktur kendali yang mengambil satu per satu anggota dari sebuah urutan (*iterable*) seperti hasil `range()`, *list*, atau *string*, lalu menjalankan blok kode untuk setiap anggota tersebut. Berbeda dengan `for` pada bahasa C, `for` pada Python tidak memiliki ekspresi inisialisasi, kondisi, dan pembaruan yang ditulis manual; variabel pencacah diisi otomatis oleh Python pada tiap iterasi. Sintaks dasarnya:

   ```python
   for variabel in urutan:
       blok_instruksi
   ```

2. Jenis-jenis Perulangan `for`

   a. `for` tunggal dengan `range()`: satu perulangan yang berjalan sejumlah anggota deret. Dipakai untuk mencetak daftar, menjumlahkan, dan sebagainya.

   b. `for` dengan kondisi di dalam blok (`if` di dalam `for`): perulangan yang di dalamnya terdapat percabangan, misalnya untuk menentukan kapan tanda `+` atau `=` dicetak pada program penjumlahan.

   c. *Nested for* (perulangan bersarang): perulangan `for` yang berada di dalam badan perulangan `for` lain. Perulangan luar biasanya mewakili **baris** dan perulangan dalam mewakili **kolom**. Jika perulangan luar berjalan `m` kali dan perulangan dalam `n` kali per putaran, total eksekusi badan dalam adalah `m × n` kali. Jika batas perulangan dalam bergantung pada variabel luar, total iterasi dihitung sebagai jumlah deret aritmetika.

3. Fungsi `range()`

   | Bentuk | Deret yang dihasilkan | Contoh |
   |---|---|---|
   | `range(stop)` | 0 sampai `stop - 1` | `range(5)` → 0, 1, 2, 3, 4 |
   | `range(start, stop)` | `start` sampai `stop - 1` | `range(1, 6)` → 1, 2, 3, 4, 5 |
   | `range(start, stop, step)` | `start`, `start + step`, ... sebelum `stop` | `range(5, 0, -1)` → 5, 4, 3, 2, 1 |

   Jumlah anggota `range(start, stop)` untuk `step = 1` adalah `stop - start`. Nilai `stop` tidak pernah ikut tercetak, sehingga untuk mencapai angka `N` dituliskan `range(1, N + 1)`.

4. Parameter `end` pada `print()`

   Secara bawaan `print()` mengakhiri keluaran dengan karakter baris baru (`\n`). Dengan `print("*", end="")` kursor tidak berpindah baris sehingga karakter dapat disambung dalam satu baris. Pemanggilan `print()` tanpa argumen dipakai untuk pindah baris setelah perulangan kolom selesai.

## 2.2 Kode Program

Bagian ini mendokumentasikan seluruh latihan praktikum perulangan `for` (Latihan 1 s.d. 3) dan perulangan bersarang (Latihan 4 s.d. 7), disusun berurutan mulai dari `perulangan_for` kemudian `nested_for`.

### 2.2.1 Program "perulangan_for1_NIM"

````python
# Meminta jumlah perulangan dari pengguna dan dikonversi ke tipe data integer
ulang_1009 = int(input("Masukkan jumlah perulangan: "))

# range(ulang) menghasilkan 0, 1, ..., ulang-1 sehingga badan perulangan berjalan sebanyak 'ulang' kali
for i_1009 in range(ulang_1009):
    # i dimulai dari 0, maka ditambah 1 agar penomoran tampil mulai dari 1
    print(f"Perulangan ke-{i_1009 + 1}")
````

**Penjelasan :**

Program menerima satu bilangan bulat `ulang_1009` lalu mengulang `print` sebanyak nilai tersebut. `range(ulang_1009)` menghasilkan deret `0` sampai `ulang_1009 - 1`, sehingga banyak iterasinya tepat `ulang_1009` kali.

**Analisis formula dan alur logika :**

- Jumlah iterasi = `ulang` (untuk `range(stop)` dengan `start = 0`).
- Nomor yang ditampilkan = `i + 1`, karena `i` bernilai 0 pada putaran pertama, sehingga hasil tercetak "Perulangan ke-1" hingga "Perulangan ke-`ulang`".
- Alur untuk input 5: `i = 0` → cetak ke-1, `i = 1` → ke-2, ..., `i = 4` → ke-5, lalu deret habis dan perulangan berhenti.

### 2.2.2 Program "perulangan_for2_NIM"

````python
ulang_1009 = int(input("Masukkan jumlah perulangan: "))

# Bagian 1: mencetak nilai pencacah apa adanya, yaitu 0 sampai ulang-1
print("Perulangan ke-0 sampai ke-", ulang_1009 - 1)

for i_1009 in range(ulang_1009):
    print(i_1009, end=" ")      # end=" " menyambung angka dalam satu baris dipisah spasi

print()                         # pindah baris

# Bagian 2: mencetak nilai pencacah + 1, yaitu 1 sampai ulang
print(f"Perulangan ke-1 sampai ke-{ulang_1009}")
for i_1009 in range(ulang_1009):
    print(i_1009 + 1, end=" ")
````

**Penjelasan :**

Program menampilkan dua deret dari perulangan `range(ulang_1009)` yang sama. Deret pertama mencetak nilai `i_1009` langsung (mulai dari 0), sedangkan deret kedua mencetak `i_1009 + 1` (mulai dari 1). Parameter `end=" "` membuat semua angka berada dalam satu baris, dan `print()` kosong dipakai untuk pindah baris di antara dua bagian.

**Analisis formula dan alur logika :**

| Putaran | `i` | Deret 1 (`i`) | Deret 2 (`i + 1`) |
|---|---|---|---|
| 1 | 0 | 0 | 1 |
| 2 | 1 | 1 | 2 |
| 3 | 2 | 2 | 3 |
| 4 | 3 | 3 | 4 |
| 5 | 4 | 4 | 5 |

- Deret 1 berakhir pada `ulang - 1`, deret 2 berakhir pada `ulang`; keduanya sama-sama berjumlah `ulang` anggota.
- Pada baris judul, `print("...ke-", ulang_1009 - 1)` memakai koma sehingga Python menyisipkan satu spasi otomatis di antara dua argumen (tampil "ke- 4"), sedangkan pada judul kedua dipakai f-string sehingga tanpa spasi tambahan ("ke-5").

### 2.2.3 Program "perulangan_for3_NIM"

````python
ulang_1009 = int(input("Masukkan jumlah perulangan: "))

# Variabel akumulator, diawali 0 sebagai nilai netral penjumlahan
jumlah_1009 = 0

for i_1009 in range(1, ulang_1009 + 1):   # i = 1, 2, ..., ulang
    print(i_1009, end=" ")
    jumlah_1009 = (jumlah_1009 + i_1009)  # menambahkan nilai i ke akumulator

    # Selama belum angka terakhir cetak tanda +, pada angka terakhir cetak tanda = beserta hasilnya
    if i_1009 < ulang_1009:
        print(" + ", end="")
    else:
        print(" = ", jumlah_1009, end="")

print()
print(f"Jumlah: {jumlah_1009}")
````

**Penjelasan :**

Program menjumlahkan bilangan bulat dari 1 sampai `ulang_1009` sambil menampilkan prosesnya. Variabel `jumlah_1009` berperan sebagai akumulator yang bertambah `i_1009` pada setiap putaran. Percabangan `if` di dalam perulangan menentukan simbol yang dicetak setelah angka: `+` bila masih ada angka berikutnya, dan `=` beserta hasil bila sudah angka terakhir.

**Analisis formula dan alur logika :**

- Hasil akhir adalah deret aritmetika 1 + 2 + ... + N yang dapat dihitung langsung dengan rumus Gauss: **S = N × (N + 1) / 2**. Untuk N = 5 diperoleh 5 × 6 / 2 = 15, sama dengan hasil program.
- Jejak akumulator untuk N = 5: 0 → 1 → 3 → 6 → 10 → 15.
- Jumlah tanda `+` yang tercetak = N - 1, karena tanda `=` menggantikan `+` pada angka terakhir.
- `range(1, ulang_1009 + 1)` dipakai agar batas atas ikut terhitung, sebab nilai `stop` pada `range` tidak pernah dijalankan.

### 2.2.4 Program "nested_for1_NIM"

````python
batas_1009 = int(input("Masukkan nilai batas: "))

# Perulangan luar: mengatur baris, i = 1 sampai batas
for i_1009 in range(1, batas_1009 + 1):

    # Perulangan dalam: mencetak titik sebanyak (batas - i) pada baris ke-i
    for j_1009 in range(1, (-1 * i_1009 + batas_1009) + 1):
        print(".", end="")

    # Setelah titik selesai, cetak nomor baris lalu pindah baris
    print(i_1009)
````

**Penjelasan :**

Program mencetak pola segitiga rata kanan. Perulangan luar menentukan nomor baris `i_1009`. Perulangan dalam mencetak karakter titik sebanyak `batas_1009 - i_1009` kali pada baris tersebut, kemudian nomor baris dicetak dan kursor pindah ke baris berikutnya melalui `print(i_1009)`.

**Analisis formula dan alur logika :**

- Batas perulangan dalam `range(1, (-i + N) + 1)` berjumlah `N - i` iterasi, sehingga jumlah titik per baris = **N - i**.
- Total titik seluruh pola = (N-1) + (N-2) + ... + 0 = **N(N - 1) / 2**. Untuk N = 5 total titiknya 10.
- Pada baris terakhir (`i = N`) perulangan dalam tidak berjalan sama sekali karena `range(1, 1)` kosong, sehingga hanya angka `N` yang dicetak.

| Baris `i` | Jumlah titik (N - i) | Keluaran |
|---|---|---|
| 1 | 4 | `....1` |
| 2 | 3 | `...2` |
| 3 | 2 | `..3` |
| 4 | 1 | `.4` |
| 5 | 0 | `5` |

### 2.2.5 Program "nested_for2_NIM"

````python
batas_1009 = int(input("Masukkan nilai batas: "))

# Perulangan luar: mengatur jumlah baris (batas baris)
for i_1009 in range(1, batas_1009 + 1):

    # Perulangan dalam: mengatur jumlah kolom pada setiap baris (batas kolom)
    for j_1009 in range(1, batas_1009 + 1):
        print("*", end="")      # cetak bintang tanpa pindah baris

    print()                     # satu baris selesai, pindah ke baris berikutnya
````

**Penjelasan :**

Program mencetak persegi bintang berukuran `batas_1009 × batas_1009`. Perulangan luar berjalan sebanyak `batas_1009` kali sebagai jumlah baris, dan di setiap barisnya perulangan dalam mencetak `batas_1009` bintang sebagai jumlah kolom. `print()` tanpa argumen setelah perulangan dalam memindahkan kursor ke baris baru.

**Analisis formula dan alur logika :**

- Total bintang = baris × kolom = **N × N = N²**. Untuk N = 5 dicetak 25 bintang.
- Perulangan dalam dijalankan ulang dari awal pada setiap putaran perulangan luar, sehingga `j` kembali bernilai 1 di tiap baris baru.
- Karena batas dalam tidak bergantung pada `i`, setiap baris memiliki panjang yang sama dan hasilnya berbentuk persegi.

### 2.2.6 Program "nested_for3_NIM"

````python
batas_1009 = int(input("Masukkan nilai batas: "))

# range(batas+1) menghasilkan 0 sampai batas, sehingga tabel berukuran (batas+1) x (batas+1)
for i_1009 in range(batas_1009 + 1):          # i sebagai nomor baris
    for j_1009 in range(batas_1009 + 1):      # j sebagai nomor kolom
        print(i_1009 + j_1009, end=" ")       # isi sel = jumlah baris dan kolom
    print()
````

**Penjelasan :**

Program membentuk tabel penjumlahan. Setiap sel pada baris `i_1009` dan kolom `j_1009` berisi hasil `i_1009 + j_1009`. Kedua perulangan memakai `range(batas_1009 + 1)` sehingga nilai dimulai dari 0 dan berakhir pada `batas_1009`.

**Analisis formula dan alur logika :**

- Ukuran tabel = (N + 1) × (N + 1), jadi untuk N = 4 terdapat 25 sel.
- Nilai sel = **i + j**, dengan nilai terkecil 0 (di pojok kiri atas) dan terbesar 2N (di pojok kanan bawah), yaitu 8 untuk N = 4.
- Setiap baris dimulai dengan nilai `i` dan bertambah satu per kolom, sehingga baris berikutnya merupakan baris sebelumnya yang digeser naik satu angka.

### 2.2.7 Program "nested_for4_NIM"

````python
tinggi_1009 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

# Validasi: tinggi harus genap
if tinggi_1009 % 2 != 0:
    print("Tinggi harus bilangan genap")
else:
    a_1009 = tinggi_1009                     # penghitung penurun, berkurang 2 tiap baris
    c_1009 = a_1009                          # posisi kolom tanda '<' sisi kiri pada baris berjalan
    lebar_1009 = (2 * tinggi_1009) - 2       # lebar pola = 2T - 2 kolom

    for i_1009 in range(1, tinggi_1009 + 1):     # perulangan luar: baris
        b_1009 = c_1009 + 1                      # posisi kolom tanda '>' pasangan '<'

        for j_1009 in range(1, lebar_1009 + 1):  # perulangan dalam: kolom

            # Baris pertama dan terakhir: bingkai atas/bawah
            if i_1009 == 1 or i_1009 == tinggi_1009:
                if j_1009 == 1 or j_1009 == lebar_1009:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris tengah
            else:
                if j_1009 == 1 or j_1009 == lebar_1009:     # pagar kiri dan kanan
                    print("|", end="")
                else:
                    if j_1009 == c_1009:                     # '<' sisi kiri
                        print("<", end="")
                    elif j_1009 == b_1009:                   # '>' sisi kiri
                        print(">", end="")
                    elif j_1009 == (lebar_1009 - c_1009):    # '<' sisi kanan (cermin)
                        print("<", end="")
                    elif j_1009 == (lebar_1009 - c_1009 + 1):  # '>' sisi kanan (cermin)
                        print(">", end="")
                    elif j_1009 > b_1009 and j_1009 < (lebar_1009 - c_1009):
                        print(".", end="")                   # isi di antara dua tanda
                    else:
                        print(" ", end="")                   # ruang kosong di luar tanda

        print()   # pindah baris

        # Memperbarui posisi tanda untuk baris berikutnya
        a_1009 -= 2
        if a_1009 <= 0:
            c_1009 = (-a_1009) + 2    # fase naik: tanda bergerak kembali menjauhi tengah
        else:
            c_1009 = a_1009           # fase turun: tanda bergerak mendekati tepi
````

**Penjelasan :**

Program mencetak pola berbingkai dengan dua pasang tanda `<>` yang bergerak simetris. Pada baris pertama dan terakhir dicetak bingkai `#====#`; pada baris tengah dicetak pagar `|` di tepi, tanda `<` dan `>` pada posisi tertentu, serta titik `.` sebagai pengisi di antara kedua tanda. Setiap karakter ditentukan oleh posisi kolom `j_1009` yang dibandingkan dengan `c_1009`, `b_1009`, dan `lebar_1009`.

**Analisis formula dan alur logika :**

- Lebar pola **W = 2T - 2** (T = tinggi), sehingga untuk T = 10 lebarnya 18 kolom.
- Posisi `<` sisi kiri = `c`, `>` sisi kiri = `c + 1`; sisi kanan merupakan cerminannya, yaitu `<` pada kolom `W - c` dan `>` pada kolom `W - c + 1`.
- Variabel `a` dimulai dari T dan turun 2 setiap baris. Selama `a > 0`, `c = a`; setelah `a ≤ 0`, `c = -a + 2`. Untuk baris tengah ke-i dapat dituliskan: `c = T - 2i + 2` bila `i ≤ T/2` dan `c = 2i - T` bila `i > T/2`.
- Jumlah titik pengisi pada tiap baris tengah = `(W - c - 1) - (c + 1) = 2T - 2c - 4`. Pada T = 10, nilai `c` berturut-turut 8, 6, 4, 2, 2, 4, 6, 8 menghasilkan titik 0, 4, 8, 12, 12, 8, 4, 0 sehingga membentuk pola kristal yang melebar ke tengah lalu menyempit lagi.
- Karena pola dibentuk per karakter lewat `if-elif-else` di dalam *nested for*, tidak diperlukan perkalian string; perulangan luar mengatur baris dan perulangan dalam mengatur kolom.

### Studi Kasus: Pola Jam Pasir Kristal Palindromik Berbingkai (*Framed Palindromic Crystal Hourglass*)
Program menerima bilangan bulat positif `N` sebagai skala jam pasir dan mencetak pola dua dimensi yang terbungkus bingkai `#====#`. Pola dibagi menjadi bingkai atas, **Fase 1** (jam pasir atas, angka `N` turun ke 1), **Fase 2** (poros titik pusat), **Fase 3** (jam pasir bawah, angka 1 naik ke `N`), dan bingkai bawah. Seluruh karakter (border, spasi, dan angka) dicetak menggunakan perulangan `for` bersarang dengan `end=""`, tanpa perkalian string. Seluruh variabel berakhiran `_1009`.

````python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_1009 = int(input("Masukkan ukuran skala jam pasir (N): "))

if n_1009 <= 0:
    print("Ukuran N harus bilangan bulat positif")
else:
    #  BINGKAI ATAS
    print("#", end="")
    for garis_1009 in range(4 * n_1009 + 5):
        print("=", end="")
    print("#")

    #  JAM PASIR ATAS (baris N - 1) 
    for baris_1009 in range(n_1009, 0, -1):
        print("|", end=" ")                                   # pagar kiri + 1 spasi padding
        for spasi_1009 in range(2 * (n_1009 - baris_1009)):   # spasi penyeimbang kiri
            print(" ", end="")
        for angka_1009 in range(baris_1009, 0, -1):           # deret mundur: baris .. 1
            print(angka_1009, end=" ")
        print("<*>", end="")                                  # poros kristal
        for angka_1009 in range(1, baris_1009 + 1):           # deret maju: 1 .. baris
            print(" ", end="")
            print(angka_1009, end="")
        for spasi_1009 in range(2 * (n_1009 - baris_1009)):   # spasi penyeimbang kanan
            print(" ", end="")
        print(" |", end="")                                   # 1 spasi padding + pagar kanan
        print()

    #  POROS TITIK PUSAT
    print("|", end="")
    for spasi_1009 in range(2 * n_1009 + 1):
        print(" ", end="")
    print("<*>", end="")
    for spasi_1009 in range(2 * n_1009 + 1):
        print(" ", end="")
    print("|")

    #  JAM PASIR BAWAH (baris 1 - N) 
    for baris_1009 in range(1, n_1009 + 1):
        print("|", end=" ")
        for spasi_1009 in range(2 * (n_1009 - baris_1009)):
            print(" ", end="")
        for angka_1009 in range(baris_1009, 0, -1):
            print(angka_1009, end=" ")
        print("<*>", end="")
        for angka_1009 in range(1, baris_1009 + 1):
            print(" ", end="")
            print(angka_1009, end="")
        for spasi_1009 in range(2 * (n_1009 - baris_1009)):
            print(" ", end="")
        print(" |", end="")
        print()

    #  BINGKAI BAWAH 
    print("#", end="")
    for garis_1009 in range(4 * n_1009 + 5):
        print("=", end="")
    print("#")
````

**Penjelasan :**

Program tugas dibangun dari tiga fase yang dibungkus dua bingkai horizontal. Semua kolom dicetak dengan `print(..., end="")` di dalam perulangan, dan `print()` tanpa argumen dipakai untuk pindah baris di akhir tiap baris pola.

**Analisis formula matematis :**

| Bagian | Formula | Untuk N = 4 |
|---|---|---|
| Panjang tanda `=` pada bingkai | 4N + 5 | 21 |
| Lebar total pola (termasuk `#` atau `\|` di kedua ujung) | 4N + 7 | 23 |
| Spasi penyeimbang kiri/kanan pada baris `b` (Fase 1 dan 3) | 2(N - b) | 0, 2, 4, 6 |
| Spasi penyeimbang kiri/kanan pada poros (Fase 2) | 2N + 1 | 9 |
| Jumlah angka per sisi pada baris `b` | b | 4, 3, 2, 1 |
| Jumlah baris keseluruhan | 1 + N + 1 + N + 1 = 2N + 3 | 11 |

Verifikasi lebar baris Fase 1: 1 (`|`) + 1 (padding) + 2(N-b) + 2b (angka mundur beserta spasi) + 3 (`<*>`) + 2b (angka maju beserta spasi) + 2(N-b) + 1 (padding) + 1 (`|`) = **4N + 7**, sama dengan lebar bingkai sehingga seluruh sisi pola rata.

**Alur logika :**

1. Bingkai atas: cetak `#`, lalu perulangan `for garis in range(4N + 5)` mencetak `=`, kemudian cetak `#`.
2. Fase 1: perulangan luar `for baris in range(N, 0, -1)` (turun dari N ke 1). Di dalamnya ada empat perulangan dalam berurutan: spasi kiri `range(2*(N-baris))`, deret mundur `range(baris, 0, -1)` yang mencetak angka lalu spasi, poros `<*>`, deret maju `range(1, baris+1)` yang mencetak spasi lalu angka, dan spasi kanan `range(2*(N-baris))`. Baris ditutup dengan ` |` dan `print()`.
3. Fase 2: dua perulangan `range(2N + 1)` untuk spasi kiri dan kanan mengapit `<*>`.
4. Fase 3: identik dengan Fase 1, tetapi perulangan luar berjalan naik `range(1, N + 1)` sehingga bentuk menjadi cermin vertikal.
5. Bingkai bawah: sama dengan bingkai atas.

Pola bersifat palindromik karena deret kiri (`b ... 1`) merupakan cermin deret kanan (`1 ... b`) terhadap poros `<*>`, dan bentuk atas merupakan cermin bawah terhadap poros titik pusat.
