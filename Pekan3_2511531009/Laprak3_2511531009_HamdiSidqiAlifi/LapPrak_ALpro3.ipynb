## 2.2 Kode Program

### 2.2.1 Program "aritmatika_NIM"


```python
angka1_1009 = int(input("input angka 1: "))
angka2_1009 = int(input("input angka 2: "))

# +
hasiladd_1009 = angka1_1009 + angka2_1009
print("\nOperator Penjumlahan")
print("Hasil: ", hasiladd_1009)

# -
hasilmin_1009 = angka1_1009 - angka2_1009
print("\nOperator Pengurangan")
print("Hasil: ", hasilmin_1009)

# *
hasiltimes_1009 = angka1_1009 * angka2_1009
print("\nOperator Perkalian")
print("Hasil: ", hasiltimes_1009)

# /
if angka2_1009 != 0:
    hasildiv_1009 = angka1_1009 / angka2_1009
    print("\nOperator Pembagian")
    print("Hasil: ", hasildiv_1009)

    hasildivv_1009 = angka1_1009 // angka2_1009 # Round floor
    print("\nOperator Pembagian Bulat")
    print("Hasil: ", hasildivv_1009)

    hasilmod_1009 = angka1_1009 % angka2_1009
    print("\nOperator Sisa Bagi")
    print("Hasil: ", hasilmod_1009)
else:
    print("Angka Tidak Boleh Bernilai Nol")
    


# Pangkat
hasilpow_1009 = angka1_1009 ** angka2_1009
print("\nOperator Pangkat")
print("Hasil: ", hasilpow_1009)
```

**Penjelasan :**

Program ini diawali dengan menerima dua input bilangan bulat dari pengguna melalui fungsi input() yang dikonversi menjadi integer menggunakan int(). 

Selanjutnya, program melakukan serangkaian operasi matematika dasar secara berurutan, yaitu 
- penjumlahan (+), 
- pengurangan (-), 
- dan perkalian (*). 

Untuk menghindari terjadinya kesalahan program berupa ZeroDivisionError saat pembagian dengan angka nol, program menerapkan struktur pengondisian if untuk memeriksa apakah variabel pembagi bernilai tidak sama dengan zero; jika kondisi terpenuhi, program mengeksekusi operasi pembagian biasa (/) yang menghasilkan tipe data float, pembagian bulat (//) yang membulatkan hasil ke bawah, dan operasi modulo (%) untuk mendapatkan sisa hasil pembagian. Namun, jika penyebut bernilai nol, program akan mengeksekusi blok else dengan menampilkan pesan peringatan ke layar. 

Terakhir, di luar blok pengondisian, program menghitung hasil perpangkatan (**) dari angka pertama terhadap angka kedua.

### 2.2.2 Program "logika_NIM"

```python
nilai1_1009 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
nilai2_1009 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true" == "true"

print("\nA1 =", nilai1_1009)
print("A2 =", nilai2_1009)

# and
hasil_1009 = nilai1_1009 and nilai2_1009
print("\nKonjungsi (AND)")
print(f"{nilai1_1009} AND {nilai2_1009} = ", hasil_1009)

# or
hasil_1009 = nilai1_1009 or nilai2_1009
print("\nDisjungsi (OR)")
print(f"{nilai1_1009} OR {nilai2_1009} = ", hasil_1009)

# NOT
hasil_1009 = not nilai1_1009
print("\nNegasi A1 (NOT)")
print(f"NOT {nilai1_1009} = ", hasil_1009)

hasil_1009 = not nilai2_1009
print("\nNegasi A2 (NOT)")
print(f"NOT {nilai2_1009} = ", hasil_1009)

# xor
hasil_1009 = nilai1_1009 != nilai2_1009
print("\nDisjungsi Ekslusif (XOR)")
print(f"{nilai1_1009} XOR {nilai2_1009} = ", hasil_1009)

```

**Penjelasan :** 

Program ini bertujuan untuk memproses dan mengevaluasi operasi logika boolean berdasarkan input teks dari pengguna.

Proses diawali dengan mengambil dua baris input berupa string "true" atau "false", kemudian merapikan format teks menggunakan metode .strip() untuk menghapus spasi liar, .lower() untuk menyeragamkan huruf kecil, serta operator == "true" untuk secara langsung mengonversinya menjadi tipe data boolean. 

Selanjutnya, program mengevaluasi kedua nilai boolean tersebut menggunakan operator logika and yang bernilai True hanya jika kedua operand bernilai benar, operator or yang bernilai True jika salah satu operand benar, dan operator not untuk membalikkan nilai boolean dari masing-masing variabel input. Untuk mengimplementasikan operasi Disjungsi Eksklusif (XOR) yang tidak memiliki kata kunci bawaan di Python, program menggunakan operator pertidaksamaan (!=), sehingga ekspresi akan mengembalikan nilai True hanya apabila kedua nilai boolean yang dibandingkan berbeda satu sama lain.


### 2.2.3 Program "perbandingan_NIM"

```python
angka1_1009 = int(input("input angka 1: "))
angka2_1009 = int(input("input angka 2: "))

# >
hasilBool_1009 =  angka1_1009 > angka2_1009
print("\nOperator Lebih Besar Dari")
print(f"{angka1_1009} > {angka2_1009} =", hasilBool_1009)

# <
hasilBool_1009 =  angka1_1009 < angka2_1009
print("\nOperator Lebih Kecil Dari")
print(f"{angka1_1009} < {angka2_1009} =", hasilBool_1009)

# >
hasilBool_1009 =  angka1_1009 >= angka2_1009
print("\nOperator Besar Sama Dari")
print(f"{angka1_1009} >= {angka2_1009} =", hasilBool_1009)

# >
hasilBool_1009 =  angka1_1009 <= angka2_1009
print("\nOperator Kecil Sama Dari")
print(f"{angka1_1009} <= {angka2_1009} =", hasilBool_1009)

# ==
hasilBool_1009 =  angka1_1009 == angka2_1009
print("\nOperator Sama")
print(f"{angka1_1009} == {angka2_1009} =", hasilBool_1009)

# !=
hasilBool_1009 =  angka1_1009 != angka2_1009
print("\nOperator Tidak Sama")
print(f"{angka1_1009} != {angka2_1009} =", hasilBool_1009)

# ext : perbandingan berantai
hasilBool_1009 = 0 < angka1_1009 <100
print("\nPerbandingan Berantai")
print(f"0 < {angka1_1009} < 100 =", hasilBool_1009)

hasilBool_1009 = 0 < angka2_1009 <100
print(f"0 < {angka2_1009} < 100 =", hasilBool_1009)
```

**Penjelasan :**

Program ini digunakan untuk mengevaluasi hubungan antara dua nilai input integer menggunakan operator relasional atau perbandingan. 

Program secara berurutan mengeksekusi dan menampilkan hasil dari operator lebih besar dari (>), lebih kecil dari (<), lebih besar sama dengan (>=), lebih kecil sama dengan (<=), sama dengan (==), dan tidak sama dengan (!=), yang mana seluruh operasi tersebut secara otomatis mengembalikan nilai bertipe boolean (True atau False). 

Pada bagian akhir program, ditunjukkan fitur perbandingan berantai khas Python melalui sintaks 0 < angka < 100, di mana ekspresi ini dievaluasi secara logis untuk memeriksa apakah nilai input berada di dalam rentang terbuka antara 0 hingga 100 tanpa perlu menggunakan operator logika tambahan secara terpisah.

### 2.2.4 Program "assignment_NIM"

```python
angka1_1009 = int(input("input angka 1: "))
angka2_1009 = int(input("input angka 2: "))

print("\nNiali angka 1 =", angka1_1009)
print("Nilai angka 2 =", angka2_1009)

# =
hasil_1009 = angka1_1009
print("\nAssigmnent Biasa")
print("Hasil =", hasil_1009)

# +=
hasil_1009 = angka1_1009
hasil_1009 += angka2_1009
print("\nAssigmnent Penambahan")
print("Hasil =", hasil_1009)

# -=
hasil_1009 = angka1_1009
hasil_1009 -= angka2_1009
print("\nAssigmnent Pengurangan")
print("Hasil =", hasil_1009)

# *=
hasil_1009 = angka1_1009
hasil_1009 *= angka2_1009
print("\nAssigmnent Perkalian")
print("Hasil =", hasil_1009)

# /=
if angka2_1009 != 0:
    hasil_1009 = angka1_1009
    hasil_1009 /= angka2_1009
    print("\nAssigmnent Pembagian")
    print("Hasil =", hasil_1009)
    
    hasil_1009 = angka1_1009
    hasil_1009 //= angka2_1009
    print("\nAssigmnent Pembagian Bulat")
    print("Hasil =", hasil_1009)

    hasil_1009 = angka1_1009
    hasil_1009 %= angka2_1009
    print("\nAssigmnent Sisa Bagi")
    print("Hasil =", hasil_1009)
else:
    print("\nPembagian tidak dapat dilakukan")
    print("Angka tidak boleh bernilai nol")
    
```

**Penjelasan :**

Program ini bertugas mendemonstrasikan penggunaan operator penugasan (assignment) sederhana dan compound assignment pada Python, diawali dengan mengambil dua input bilangan bulat dari pengguna serta menampilkan kembali nilai awalnya ke layar.

 Operasi dimulai dengan assignment biasa (=) untuk mengisi nilai variabel hasil dari angka pertama, dilanjutkan dengan compound assignment penambahan (+=), pengurangan (-=), dan perkalian (*=) yang masing-masing secara ringkas menggabungkan operasi aritmatika dengan pembaruan nilai variabel. 
 
 Sebelum menjalankan operasi pembagian, program menggunakan struktur pengondisian if untuk memverifikasi bahwa angka kedua tidak bernilai nol guna mencegah error runtime; jika aman, program mengeksekusi compound assignment pembagian (/=) yang secara otomatis mengubah tipe data variabel hasil menjadi float, pembagian bulat (//=), serta modulo (%=). Jika input angka kedua adalah nol, program akan melompati operasi pembagian dan langsung mencetak pesan bahwa pembagian tidak dapat dilakukan.


### 2.2.5 Program "bitwise_NIM"

```python
print("==============================")
print("     2. OPERATOR BITWISE      ")
print("==============================")

# input angka
angka1_1009 = int(input("input angka 1: "))
angka2_1009 = int(input("input angka 2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_1009, "| biner =", bin(angka1_1009))
print("angka1 =", angka2_1009, "| biner =", bin(angka2_1009))

# Bitwise AND
hasil_1009 = angka1_1009 & angka2_1009
print("\nBitwise AND (&)")
print(angka1_1009, "&", angka2_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))

# Bitwise OR
hasil_1009 = angka1_1009 | angka2_1009
print("\nBitwise OR (|)")
print(angka1_1009, "|", angka2_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))

# Bitwise XOR
hasil_1009 = angka1_1009 ^ angka2_1009
print("\nBitwise XOR (^)")
print(angka1_1009, "^", angka2_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))

# Bitwise NOT
hasil_1009 = ~angka1_1009
print("\nBitwise NOT (~)")
print("~", angka1_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))

 
# Left Shift
jumlah_geser_1009 = int(input("\nMasukkan jumlah pergeseran bit:"))

hasil_1009 = angka1_1009 << jumlah_geser_1009
print("\nBitwise Left Shift (<<)")
print(angka1_1009, "<<", jumlah_geser_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))
 
# Left Shift
hasil_1009 = angka1_1009 >> jumlah_geser_1009
print("\nBitwise Right Shift (>>)")
print(angka1_1009, ">>", jumlah_geser_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))
 
```

**Penjelasan :**

Program ini mendemonstrasikan manipulasi data pada tingkat biner dengan terlebih dahulu membaca dua input integer dan menampilkan representasi desimal beserta nilai binernya menggunakan fungsi bawaan bin(). 

Program kemudian menjalankan dan menampilkan hasil dari operasi Bitwise AND (&) yang menghasilkan bit 1 jika kedua bit bernilai 1, Bitwise OR (|) yang menghasilkan bit 1 jika salah satu bit bernilai 1, Bitwise XOR (^) yang menghasilkan bit 1 jika kedua bit berbeda, serta Bitwise NOT (~) yang membalikkan seluruh bit dan dihitung secara matematis menggunakan rumus $-(x + 1)$. 

Setiap hasil operasi ditampilkan dalam format nilai desimal dan string biner 8-bit (zero-padded) menggunakan fungsi format(x, "08b"). 

Di bagian akhir, program meminta input jumlah pergeseran bit dari pengguna untuk menjalankan operasi Left Shift (<<) yang menggeser bit ke kiri atau setara dengan perkalian kelipatan $2^n$, serta Right Shift (>>) yang menggeser bit ke kanan atau setara dengan pembagian bulat kelipatan $2^n$.


### 2.2.6 Program "lainnya_NIM"

```python
print("==============================")
print("   1. OPERATOR KEANGGOTAAN    ")
print("==============================")

# input data
input_data_1009 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# input mmenjadi list integer
data_1009 = [int(angka.strip()) for angka in input_data_1009.split(",")]

nilai_dicari_1009 = int(input("masukkan angka yang ingin dicari: "))

# in
hasil_1009 = nilai_dicari_1009 in data_1009
print("\nOperator Keanggotaan IN")
print(f"{nilai_dicari_1009} in {data_1009} = ", hasil_1009)

# operator not in
hasil_1009 = nilai_dicari_1009 not in data_1009
print("\nOperator Keanggotaan NOT IN")
print(f"{nilai_dicari_1009} in {data_1009} = ", hasil_1009)

print("\n=====================================")
print("         2. OPERATOR IDENTITAS       ")
print("=====================================")

# objek1 menggunakan list dari input pengguna
objek1_1009 =data_1009

# objek2 merujuk pada objek yang sama dengan objek1
objek2_1009 = objek1_1009

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_1009 = data_1009.copy()

print("objek1 =", objek1_1009)
print("objek2 =", objek2_1009)
print("objek3 =", objek3_1009)

# Operator is
hasil_1009 = objek1_1009 is objek2_1009
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_1009)

# Operator is not
hasil_1009 = objek1_1009 is not objek3_1009
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_1009)

# membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 == objek3 =", objek1_1009 == objek3_1009)
print("objek1 == objek3 =", objek1_1009 is objek3_1009)
```

**Penjelasan:**

Program ini terbagi menjadi dua bagian utama yang membahas operator keanggotaan dan operator identitas. 

Pada bagian keanggotaan, program menerima sekumpulan angka berbasis string yang dipisahkan oleh koma, kemudian mengonversinya menjadi `list of integers` memanfaatkan list comprehension, dilanjutkan dengan menguji keberadaan sebuah angka acuan di dalam list tersebut menggunakan operator `in` dan `not in`. Pada bagian identitas, program membuat variabel objek2 sebagai alias yang menunjuk ke alamat memori yang sama dengan `objek1`, serta variabel `objek3` yang dibuat menggunakan metode `.copy()` sehingga menghasilkan objek baru dengan isi nilai yang sama tetapi memiliki alamat memori berbeda. Pengujian dengan operator `is` dan `is not` membuktikan keberadaan identitas memori tersebut, di mana bagian penutup program secara tegas memperlihatkan perbedaan antara perbandingan nilai (`==`) yang bernilai `True` karena isi list sama, dengan perbandingan identitas (`is`) yang bernilai `False` karena keduanya bertempat di lokasi memori yang berbeda.
