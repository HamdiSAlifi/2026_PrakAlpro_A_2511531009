## 2.1 Dasar Teori
Beberapa dasar teori yang meliputi paraktikum ini adaalh:

1. Pengertian Tipe Data.

    Tipe data adalah jenis nilai yang dapat disimpan dan diproses oleh program. Setiap tipe data memiliki ukuran memori, operasi, dan batasan nilai yang berbeda. Python secara otomatis mendeteksi tipe data berdasarkan nilai yang diberikan (dynamic typing).

2. Tipe Data Dasar Python
    - Boolean: Tipe data logika yang hanya memiliki dua nilai, yaitu True and False. Biasanya digunakan untuk mengambil keputusan.
    - Integer: Bilangan bulat tanpa desimal.
    - Float: Bilangan desimal
    - String: Teks atau rangkaian karakter.
    - Complex: Bilangan kompleks dengan bagian real dan imajiner.

## 2.2 Kode Program

### 2.2.1 Program "numerik1_NIM"


```python
# Create a variable with integer value
a_1009 = 100
print("the type of variable having value ", a_1009, "is ", type(a_1009))

#
b_1009 = 10.2345
print("the type of varialbe having value ", b_1009, "is ", type(b_1009))

#
c_1009 = 3j
print("the type of varialbe having value ", c_1009, "is ", type(c_1009))
```

**Penjelasan :**
Dengan menggunakan `type(variable)`, kita dapat mnegetahui tipe apa sebuah variable.

### 2.2.2 Program "String_NIM"

```python
nama_1009 = "Wahyu"
jenis_kelamin_1009 = 'L'
alamat_1009 = """
    Kampus Unand
    Kec. Pauh
    Kota Padang
"""
agama_1009 = "Islam"
print("Nama saya adalah             : ", nama_1009)
print("Jenis kelamin saya adalah    : ", jenis_kelamin_1009)
print("Alamat saya adalah           : ", alamat_1009)
print("Agama saya adalah    : ", agama_1009)
```

**Penjelasan :** Dengan menggunakan (") atau (') kita dapat memamsukkan tipe data _string_ kedalam suatu variable

### 2.2.3 Program "Boolean_NIM"

```python
is_lulus_1009 = True
is_cumlaude_1009 = True

nilai_1009 = 85
batas_lulus_1009 = 75

status_kelulusan_1009 = nilai_1009 >= batas_lulus_1009

print("=== CHECK STASTUS KELULUSAN ===")
print("Nilai        : ", nilai_1009)
print("Apakah Lulus : ", status_kelulusan_1009)
if is_lulus_1009 and is_cumlaude_1009:
    print("Selamat anda lulus dengan predikat Cum Laude")
```

**Penjelasan :** True adalah nilai benar, dan False adalah nilai salah, jika diaplikasikan pada operator >, <, ==, != atau yang lainnya, tipe data ini dapat berperan sebagai operand(alamat) yang akan digunakan.

### 2.2.4 Program "Konstanta_NIM"

```python
from typing import final
PI_1009: final = 3.14

print("pi: %f" % (PI_1009))
jari_1009 = float(input("masukkan nilai jari-jari : "))
luas_1009 = PI_1009 * jari_1009 * jari_1009

print("Luas lingkaran dengan jari jari %f adalah %f" % (jari_1009, luas_1009))
```

**Penjelasan :** Dengan menggunakan library tertentu pada python kita dapat menciptakan sebuah kalkulator untuk mencari luas lingkaran. PErtama, prgram aakn emndefinisikan nilai [o sebagai konstanta menggunakan `final`] dari modul `typing`. Lalu program meminta pengguna memasukkan nilai jari-jari, mnegubahnya menjadi flaot, menghitung luas dengan rumus $\pi$ $\times$ $r^2$ , dan menampilkan hasilnya. 


### 2.2.5 Program "Input_NIM"

```python
panjang_1009 = input("Masukkan nilai panjang  : ")
lebar_1009 = input("Masukkan nilai lebar    : ")

luas_1009 = int(panjang_1009) * int(lebar_1009)

print("Luas : ", luas_1009)
```

**Penjelasan :** Dengan menggunakan input, python akan mendeteksi secara ototmatis tipe data yang diinputkan pada contoh program diatas, yang mana akan di rumabh menajdi tipe data integer dengan menggunakan  fungsi int(). Kemudian dengan memanfaatkan fungsi tersebut kita dapat menciptakan sebuah pengecualian untuk tidak terinputnya tipe data lain ke dalam variable tersebut. Jika variabel yang diinputkan adalah double/float, maka akan di sederhanakan menajdi bentuk integer.
