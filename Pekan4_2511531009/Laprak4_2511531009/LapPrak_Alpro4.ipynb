## 2.1 Dasar Teori
Beberapa dasar teori yang meliputi praktikum ini adalah:

1. Pengertian Struktur Percabangan.

    Struktur Percabangan (_conditional statement_) adalah mekanisme dalam pemrograman yang memungkinkan program untuk memilih dan menjalankan blok kode tertentu berdasarkan hasil evaluasi suatu kondisi. Kondisi tersebut dievaluasi sebagai nilai boolean, yaitu True atau False.


2. Jenis-jenis Struktur Percabangan di Python
    - if: Menjalankan blok kode hanya jika kondisi bernilai True.
    - if-else: Dua jalur eksekusi, untuk kondisi True dan satu untuk False.
    - if-elif-else: Mengevaluasi beberapa kondisi (lebih dari dua) secara berurutan.
    - match-case: Memilih blok kode berdasarkan pencocokan nilai

## 2.2 Kode Program

### 2.2.1 Program "if_NIM"


```python
# Menginisialisasi variabel untuk dinput yang otomatis hanya di parse dalam bentuk tipe data float
ipk_1009 = float(input("Input IPK anda: "))

# pengecekan kondisi variabel input, jika variabel besar dari nilai 2.75 maka akan melaksanakan aksi: mencetak pesan anda lulus dengan append  variabel input sebelumnya dengan tipe data string
if ipk_1009 > 2.75:
    print("Anda lulus Sangat Memuaskan dengan IPK " + str(ipk_1009))

# Mencetak pesan program selesai
print("Program Selesai")
```

**Penjelasan :**

Program ini mencetak pesan berdasarkan nilai IPK yang diinput. Jika IPK lebih dari 2.75, program mencetak pesan lulus beserta nilai IPK-nya. Kalau tidak, program hanya mencetak pesan selesai.

### 2.2.2 Program "if2_NIM"

```python
# Menginisialisasi variabel untuk dinput yang otomatis hanya di parse dalam bentuk tipe data float
ipk_1009 = float(input("Input IPK anda: "))

# pengecekan kondisi variabel input, jika variabel besar dari nilai 2.75 maka akan melaksanakan aksi: mencetak pesan anda lulus dengan append  variabel input sebelumnya dengan tipe data string
if ipk_1009 > 2.75:
    print("Anda lulus Sangat Memuaskan dengan IPK " + str(ipk_1009))
else: # Jika tidak maka program mencetak pesan tidak lulus
    print("Anda tidak lulus")
    
# Mencetak pesan program selesai
print("Program Selesai")
```

**Penjelasan :** 

Program bekerja seperti `if_NIM.py` namun, terdapat kondisi jika `ipk > 2.75` bernilai false maka program akan mencetak pesan tidak lulus sebelum pesan program selesai

### 2.2.3 Program "multi_if1_NIM"

```python
# variabel umur adalah input yang diparse menjadi data type integer
umur_1009 = int(input("Input umur anda: "))

sim_1009 = input("apakah anda sudah punya SIM C? (y/n): ")

if (umur_1009 >= 17) and sim_1009 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")
    
if (umur_1009 >= 17) and sim_1009 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
    
if (umur_1009 < 17) and sim_1009 == 'y':
    print("Anda belum cukup umur punya SIM")
    
if (umur_1009 < 17) and sim_1009 != 'y':
    print("Anda belum cukup umur bawa motor")
```

**Penjelasan :**

  Program ini menerima dua input yaitu umur dan kepemilikan SIMC. Terdapat empat blok if independen yang masing-masing dievaluasi secara terpisah. Setiap kondisi menggabungkan dua syarat menggunakan operator logika and, sehingga hanya satu blok yang akan mencetak output sesuai kombinasi kondisi yang terpenuhi.

### 2.2.4 Program "if_elif_else1_NIM"

```python
# Mengambil karakter pertama dari input untuk memastikan hanya satu karakter yang dibaca
umur_1009 = int(input("Input umur anda: "))
sim_1009 = input("apakah anda sudah punya SIM C? (y/n): ")[0]

if (umur_1009 >= 17) and sim_1009 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")

elif umur_1009 >= 17 and sim_1009 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
    
elif umur_1009 < 17 and sim_1009 == 'y':
    print("Anda belum cukup umur punya SIM")
    
else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")
    
print("program selesai")
```

**Penjelasan :**

Program ini memiliki fungsi yang sama dengan multi_if1, namun menggunakan struktur if-elif-else. Perbedaan utamanya adalah begitu satu kondisi bernilai True dan bloknya dieksekusi, evaluasi kondisi berikutnya langsung dihentikan. Ini membuat program lebih efisien karena tidak perlu mengevaluasi semua kondisi jika salah satu sudah terpenuhi. Selain itu, `[0]` pada baris input `sim` mengambil karakter pertama saja dari string yang dimasukkan.


### 2.2.5 Program "multi_if2_NIM"

```python
total_belanja_1009 = float(input("Masukkan total belanja (Rp): "))

input_member_1009 = input("Apakah Anda member? (y/n): ").strip().lower()
is_member_1009 = input_member_1009 in ['y', "ya"]

input_promo_1009 = input("Apakah kode promo valid? (y/n): ").strip().lower()
kode_promo_valid_1009 = input_promo_1009 in ["y", "ya"]

total_diskon_persen_1009 = 0

if total_belanja_1009 > 100000:
    total_diskon_persen_1009 += 10

if is_member_1009:
    total_diskon_persen_1009 += 5
    
if kode_promo_valid_1009:
    total_diskon_persen_1009 += 15
    
nominal_diskon_1009 = total_belanja_1009 * (total_diskon_persen_1009 / 100)
total_bayar_1009 = total_belanja_1009 - nominal_diskon_1009

print("\n [ RINCIAN PEMBAYARAN ] ")
print(f"Total Diskon : {total_diskon_persen_1009}% (Rp{nominal_diskon_1009:,.0f})")
print(f"Total Bayar  : Rp{total_bayar_1009:,.0f}")

print(f"Total diskon yang anda dapatkan: Rp{total_diskon_persen_1009}%")
```

**Penjelasan :**

Program ini mensimulasikan sistem diskon toko dengan tiga
  kondisi if independen yang dievaluasi satu per satu. Diskon
  dapat terakumulasi dari tiga sumber secara bersamaan: total
  belanja di atas Rp100.000 (10%), status member (5%), dan kode
  promo valid (15%). Karena ketiga blok if berdiri sendiri, semua
  kondisi yang terpenuhi akan menjumlahkan persentase diskon ke
  dalam total_diskon_persen_1009 menggunakan operator +=.

### 2.2.6 Program "nilai_NIM"

```python
nilai_1009 = int(input("Masukkan Nilai: "))

if nilai_1009 >= 81:
   print("A") 
elif nilai_1009 >= 70:
   print("B") 
elif nilai_1009 >= 60:
   print("C") 
elif nilai_1009 >= 50:
   print("D") 
else: 
   print("E") 
```

**Penjelasan:**

Program ini mengonversi nilai angka menjadi nilai huruf  menggunakan struktur if-elif-else. Kondisi dievaluasi dari atas
  ke bawah secara berurutan. Nilai >= 81 mendapat A, >= 70
  mendapat B, >= 60 mendapat C, >= 50 mendapat D, dan selain itu
  mendapat E. Karena menggunakan elif, evaluasi berhenti begitu
  kondisi pertama yang cocok ditemukan.

### 2.2.7 Program "match_NIM"

```python
bulan_1009 = int(input("Masukkan Angka Bulan (1 - 12): "))

match bulan_1009: 
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan:**

Program ini menerima input angka bulan lalu mencocokkannya menggunakan `match-case`. Setiap `case` mewakili satu angka dari 1 sampai 12 dan mencetak nama bulan yang sesuai. Jika input di luar rentang tersebut, `case _` sebagai wildcard akan menangkap nilai tersebut dan mencetak pesan angka tidak valid.

## 2.3 Tugas

### Studi Kasus: Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana
Sebuah wahana rekreasi petualangan membutuhkan sistem loket berbasis konsol untuk menentukan jenis wahana, memverifikasi kelaikan pengunjung dengan kondisi majemuk, menghitung diskon akumulatif menggunakan multi-if terpisah, serta mencetak audit status operasional.

```python
# Program: Sistem Loket Terpadu Alpro Adventure Park
# NIM    : 2511531009

# BAGIAN 1: Input Data Pengunjung
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

nama_1009         = input("Masukkan Nama Pengunjung        : ")
umur_1009         = int(input("Input umur anda                 : "))
# Hanya ambil karakter pertama agar input 'ya' atau 'tidak' tetap terbaca konsisten
sim_1009          = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]

print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_1009        = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_1009 = int(input("Masukkan jumlah tiket           : "))
input_member_1009 = input("Apakah Anda member? (y/t)       : ").strip().lower()
input_promo_1009  = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# Percabangan 1: if tunggal - validasi kelogisan jumlah tiket
if jumlah_tiket_1009 <= 0:
    print("\nPeringatan: Kuota tiket tidak valid! Jumlah tiket harus lebih dari 0.")

# Inisialisasi variabel wahana sebelum diisi oleh match-case
nama_wahana_1009  = ""
harga_satuan_1009 = 0
paket_valid_1009  = True

# Percabangan 2: match-case - pemilihan wahana dan harga satuan
match paket_1009:
    case 1:
        nama_wahana_1009  = "Wahana Safari Rimba"
        harga_satuan_1009 = 50000
    case 2:
        nama_wahana_1009  = "Wahana Arung Jeram"
        harga_satuan_1009 = 75000
    case 3:
        nama_wahana_1009  = "Wahana Motor ATV Ekstrim"
        harga_satuan_1009 = 120000
    case 4:
        nama_wahana_1009  = "Wahana Roller Coaster Kilat"
        harga_satuan_1009 = 100000
    case 5:
        nama_wahana_1009  = "Wahana All-Access VIP"
        harga_satuan_1009 = 220000
    case _:
        # Input di luar 1-5 tidak valid, eksekusi transaksi dihentikan
        print("\nPaket wahana tidak valid!")
        paket_valid_1009 = False

if paket_valid_1009 and jumlah_tiket_1009 > 0:

    print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

    # Percabangan 3: if-elif-else - validasi izin kendali wahana
    # Paket 3 memerlukan pengecekan SIM C karena melibatkan kendaraan bermotor
    if paket_1009 == 3:

        if umur_1009 >= 17 and sim_1009 == 'y':
            status_akses_1009 = "Anda sudah dewasa dan boleh mengendarai ATV sendiri."

        elif umur_1009 >= 17 and sim_1009 != 'y':
            status_akses_1009 = "Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur)."

        elif umur_1009 < 17 and sim_1009 == 'y':
            status_akses_1009 = "Identitas tidak valid: Belum cukup umur memiliki SIM."

        else:
            status_akses_1009 = "Anda belum cukup umur dan tidak boleh bawa motor ATV."

    else:
        # Untuk paket selain ATV, syarat minimum hanya usia 10 tahun
        if umur_1009 >= 10:
            status_akses_1009 = "Usia Anda memenuhi syarat untuk wahana ini."
        else:
            status_akses_1009 = "Maaf, usia Anda belum memenuhi syarat minimum (10 tahun) untuk wahana ini."

    print(f"Status Akses: {status_akses_1009}")

    # Hitung subtotal sebelum diskon diterapkan
    subtotal_1009             = harga_satuan_1009 * jumlah_tiket_1009
    total_diskon_persen_1009  = 0

    # Percabangan 4: Multi-if terpisah - akumulasi diskon bertingkat
    # Setiap blok dievaluasi sendiri sehingga diskon dapat menumpuk
    if subtotal_1009 >= 200000:
        total_diskon_persen_1009 += 10

    if input_member_1009 in ['y', 'ya']:
        total_diskon_persen_1009 += 5

    if input_promo_1009 in ['y', 'ya']:
        total_diskon_persen_1009 += 15

    if jumlah_tiket_1009 >= 5:
        total_diskon_persen_1009 += 5

    nominal_diskon_1009 = subtotal_1009 * (total_diskon_persen_1009 / 100)
    total_bayar_1009    = subtotal_1009 - nominal_diskon_1009

    # Percabangan 5: if-else - audit total bayar dan catatan layanan 
    if total_bayar_1009 > 300000:
        catatan_layanan_1009 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    else:
        catatan_layanan_1009 = "Terima kasih telah berkunjung."

    print("\n--- Rincian Pembayaran ---")
    print(f"Wahana           : {nama_wahana_1009}")
    print(f"Nama Pengunjung  : {nama_1009}")
    print(f"Subtotal Belanja : Rp {subtotal_1009:,.0f}")
    print(f"Total Diskon     : {total_diskon_persen_1009}% (Rp {nominal_diskon_1009:,.0f})")
    print(f"Total Bayar      : Rp {total_bayar_1009:,.0f}")
    print(f"Catatan Layanan  : {catatan_layanan_1009}")

print("\nProgram Selesai")
```

**Penjelasan:**

Program ini adalah sistem loket sederhana yang menggunakan semua jenis percabangan dari modul Pekan 4. Input berupa nama, umur, SIM C, pilihan paket, jumlah tiket, status member, dan kode promo. `if` tunggal mengecek apakah jumlah tiket valid, `match-case` menentukan wahana dan harga sesuai pilihan, dan `case _` menangkap input yang tidak sesuai. Khusus Paket 3, `if-elif-else` dengan `and` dan `!=` mengecek izin pengendara ATV, sementara paket lain cukup dicek umurnya pakai `if-else`. Empat blok `if` terpisah mengakumulasi diskon dari masing-masing sumber, dan `if-else` terakhir menentukan apakah pengunjung dapat souvenir gratis atau tidak.

# KESIMPULAN

Berdasarkan praktikum mengenai struktur percabangan pada bahasa pemrograman Python yang telah dilaksanakan, dapat disimpulkan bahwa:

1. Python menyediakan empat bentuk utama struktur percabangan, yaitu `if` tunggal, `if-else`, `if-elif-else`, dan `match-case`. Masing-masing memiliki kegunaan yang berbeda tergantung pada jumlah kondisi yang perlu dievaluasi dan kompleksitas logika program.

2. Pernyataan `if` tunggal digunakan ketika hanya ada satu kondisi yang perlu diperiksa tanpa jalur alternatif. Pernyataan `if-else` menyediakan dua jalur eksekusi, sementara `if-elif-else` menangani lebih dari dua kemungkinan kondisi secara berurutan dan berhenti di kondisi pertama yang bernilai `True`.

3. Penggunaan beberapa blok `if` yang berdiri sendiri (multi-if terpisah) berbeda secara fundamental dengan `if-elif-else`. 4. Pernyataan `match-case` yang tersedia sejak Python 3.10 menyediakan cara yang lebih bersih dan mudah dibaca untuk menangani percabangan berbasis nilai tetap. Default `case _` berfungsi sebagai penangkap default apabila tidak ada case yang cocok dengan nilai yang diberikan.

5. Operator logika `and`, `or`, dan `not` berperan penting dalam membentuk kondisi majemuk di dalam struktur percabangan. Kombinasi operator logika dengan operator perbandingan memungkinkan program mengevaluasi lebih dari satu syarat sekaligus dalam satu pernyataan `if`, sehingga program dapat menghasilkan respons yang lebih tepat terhadap berbagai kombinasi input pengguna.
