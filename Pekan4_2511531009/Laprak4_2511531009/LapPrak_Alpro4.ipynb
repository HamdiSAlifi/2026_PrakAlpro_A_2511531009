## 1.1 Latar Belakang

Dalam dunia pemrograman komputer, operator merupakan simbol khusus yang digunakan untuk memproses, memanipulasi, dan mengoperasikan nilai atau variabel (operand). Pemrograman tidak hanya berfokus pada penyimpan data dalam memori, melainkan juga bagaimana data tersebut diolah secara logis dan matematis. Oleh karena itu, pemahaman mendalam mengenai operator menjadi fondasi utama sebelum membangun struktur alur kontrol, fungsi, maupun algoritma perangkat lunak yang lebih kompleks.

Bahasa pemrograman Python menyediakan tujuh kelompok operator utama, yaitu operator aritmatika, assignment, perbandingan, logika, bitwise, keanggotaan, dan identitas. Setiap kelompok operator memiliki aturan sintaks, jenis operand, serta perilaku yang berbeda-beda. Sebagai contoh, operator pembagian biasa pada Python secara otomatis selalu mengembalikan tipe data float, sedangkan pembagian bulat membulatkan hasil ke bawah. Ketidaktelitian dalam memilih jenis operator, seperti mengabaikan potensi pembagian dengan angka nol yang dapat memicu runtime error (ZeroDivisionError), dapat menyebabkan program berhenti secara tidak normal.

Selain operasi numerik dan logika dasar, pemrosesan data tingkat lanjut pada Python juga melibatkan pemahaman mengenai manipulasi biner serta manajemen memori. Operator bitwise memungkinkan pemrosesan langsung pada level bit menggunakan sistem representasi Two's Complement, sedangkan operator keanggotaan (in / not in) menyederhanakan pencarian elemen dalam struktur data seperti list. Lebih jauh lagi, salah satu aspek krusial yang sering menjadi kekeliruan adalah membedakan antara perbandingan nilai dengan perbandingan identitas objek di memori (is). Dua variabel dapat memiliki isi data yang sama, namun belum tentu merujuk pada alamat memori yang identik. Melalui praktikum Pekan 3 ini, pengujian seluruh kelompok operator secara langsung diharapkan dapat memberikan pemahaman yang komprehensif mengenai mekanisme kerja, tipe data kembalian, serta penanganan error pada Python.

### 2.2.1 Program "aritmatika_NIM"


````python
angka1_1009 = int(input("input angka 1: "))
angka2_1009 = int(input("input angka 2: "))

# +
hasiladd_1009 = angka1_1009 + angka2_1009
print("\nOperator Penjumlahan")
print("Hasil: ", hasiladd_1009)

# -
hasilmin_1009 = angka1_1009 - angka2_1009
print("\nOperator Pengurangan")
print("Hasil: ", hasilmin_1009)

# *
hasiltimes_1009 = angka1_1009 * angka2_1009
print("\nOperator Perkalian")
print("Hasil: ", hasiltimes_1009)

# /
if angka2_1009 != 0:
    hasildiv_1009 = angka1_1009 / angka2_1009
    print("\nOperator Pembagian")
    print("Hasil: ", hasildiv_1009)

    hasildivv_1009 = angka1_1009 // angka2_1009 # Round floor
    print("\nOperator Pembagian Bulat")
    print("Hasil: ", hasildivv_1009)

    hasilmod_1009 = angka1_1009 % angka2_1009
    print("\nOperator Sisa Bagi")
    print("Hasil: ", hasilmod_1009)
else:
    print("Angka Tidak Boleh Bernilai Nol")
    


# Pangkat
hasilpow_1009 = angka1_1009 ** angka2_1009
print("\nOperator Pangkat")
print("Hasil: ", hasilpow_1009)
````

**Penjelasan :**

Program ini diawali dengan menerima dua input bilangan bulat dari pengguna melalui fungsi input() yang dikonversi menjadi integer menggunakan int(). 

Selanjutnya, program melakukan serangkaian operasi matematika dasar secara berurutan, yaitu 
- penjumlahan (+), 
- pengurangan (-), 
- dan perkalian (*). 

Untuk menghindari terjadinya kesalahan program berupa ZeroDivisionError saat pembagian dengan angka nol, program menerapkan struktur pengondisian if untuk memeriksa apakah variabel pembagi bernilai tidak sama dengan zero; jika kondisi terpenuhi, program mengeksekusi operasi pembagian biasa (/) yang menghasilkan tipe data float, pembagian bulat (//) yang membulatkan hasil ke bawah, dan operasi modulo (%) untuk mendapatkan sisa hasil pembagian. Namun, jika penyebut bernilai nol, program akan mengeksekusi blok else dengan menampilkan pesan peringatan ke layar. 

Terakhir, di luar blok pengondisian, program menghitung hasil perpangkatan (**) dari angka pertama terhadap angka kedua.

### 2.2.2 Program "logika_NIM"

````python
nilai1_1009 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
nilai2_1009 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true" == "true"

print("\nA1 =", nilai1_1009)
print("A2 =", nilai2_1009)

# and
hasil_1009 = nilai1_1009 and nilai2_1009
print("\nKonjungsi (AND)")
print(f"{nilai1_1009} AND {nilai2_1009} = ", hasil_1009)

# or
hasil_1009 = nilai1_1009 or nilai2_1009
print("\nDisjungsi (OR)")
print(f"{nilai1_1009} OR {nilai2_1009} = ", hasil_1009)

# NOT
hasil_1009 = not nilai1_1009
print("\nNegasi A1 (NOT)")
print(f"NOT {nilai1_1009} = ", hasil_1009)

hasil_1009 = not nilai2_1009
print("\nNegasi A2 (NOT)")
print(f"NOT {nilai2_1009} = ", hasil_1009)

# xor
hasil_1009 = nilai1_1009 != nilai2_1009
print("\nDisjungsi Ekslusif (XOR)")
print(f"{nilai1_1009} XOR {nilai2_1009} = ", hasil_1009)

````

**Penjelasan :** 

Program ini bertujuan untuk memproses dan mengevaluasi operasi logika boolean berdasarkan input teks dari pengguna.

Proses diawali dengan mengambil dua baris input berupa string "true" atau "false", kemudian merapikan format teks menggunakan metode .strip() untuk menghapus spasi liar, .lower() untuk menyeragamkan huruf kecil, serta operator == "true" untuk secara langsung mengonversinya menjadi tipe data boolean. 

Selanjutnya, program mengevaluasi kedua nilai boolean tersebut menggunakan operator logika and yang bernilai True hanya jika kedua operand bernilai benar, operator or yang bernilai True jika salah satu operand benar, dan operator not untuk membalikkan nilai boolean dari masing-masing variabel input. Untuk mengimplementasikan operasi Disjungsi Eksklusif (XOR) yang tidak memiliki kata kunci bawaan di Python, program menggunakan operator pertidaksamaan (!=), sehingga ekspresi akan mengembalikan nilai True hanya apabila kedua nilai boolean yang dibandingkan berbeda satu sama lain.


### 2.2.3 Program "perbandingan_NIM"

````python
angka1_1009 = int(input("input angka 1: "))
angka2_1009 = int(input("input angka 2: "))

# >
hasilBool_1009 =  angka1_1009 > angka2_1009
print("\nOperator Lebih Besar Dari")
print(f"{angka1_1009} > {angka2_1009} =", hasilBool_1009)

# <
hasilBool_1009 =  angka1_1009 < angka2_1009
print("\nOperator Lebih Kecil Dari")
print(f"{angka1_1009} < {angka2_1009} =", hasilBool_1009)

# >
hasilBool_1009 =  angka1_1009 >= angka2_1009
print("\nOperator Besar Sama Dari")
print(f"{angka1_1009} >= {angka2_1009} =", hasilBool_1009)

# >
hasilBool_1009 =  angka1_1009 <= angka2_1009
print("\nOperator Kecil Sama Dari")
print(f"{angka1_1009} <= {angka2_1009} =", hasilBool_1009)

# ==
hasilBool_1009 =  angka1_1009 == angka2_1009
print("\nOperator Sama")
print(f"{angka1_1009} == {angka2_1009} =", hasilBool_1009)

# !=
hasilBool_1009 =  angka1_1009 != angka2_1009
print("\nOperator Tidak Sama")
print(f"{angka1_1009} != {angka2_1009} =", hasilBool_1009)

# ext : perbandingan berantai
hasilBool_1009 = 0 < angka1_1009 <100
print("\nPerbandingan Berantai")
print(f"0 < {angka1_1009} < 100 =", hasilBool_1009)

hasilBool_1009 = 0 < angka2_1009 <100
print(f"0 < {angka2_1009} < 100 =", hasilBool_1009)
````

**Penjelasan :**

Program ini digunakan untuk mengevaluasi hubungan antara dua nilai input integer menggunakan operator relasional atau perbandingan. 

Program secara berurutan mengeksekusi dan menampilkan hasil dari operator lebih besar dari (>), lebih kecil dari (<), lebih besar sama dengan (>=), lebih kecil sama dengan (<=), sama dengan (==), dan tidak sama dengan (!=), yang mana seluruh operasi tersebut secara otomatis mengembalikan nilai bertipe boolean (True atau False). 

Pada bagian akhir program, ditunjukkan fitur perbandingan berantai khas Python melalui sintaks 0 < angka < 100, di mana ekspresi ini dievaluasi secara logis untuk memeriksa apakah nilai input berada di dalam rentang terbuka antara 0 hingga 100 tanpa perlu menggunakan operator logika tambahan secara terpisah.

### 2.2.4 Program "assignment_NIM"

````python
angka1_1009 = int(input("input angka 1: "))
angka2_1009 = int(input("input angka 2: "))

print("\nNiali angka 1 =", angka1_1009)
print("Nilai angka 2 =", angka2_1009)

# =
hasil_1009 = angka1_1009
print("\nAssigmnent Biasa")
print("Hasil =", hasil_1009)

# +=
hasil_1009 = angka1_1009
hasil_1009 += angka2_1009
print("\nAssigmnent Penambahan")
print("Hasil =", hasil_1009)

# -=
hasil_1009 = angka1_1009
hasil_1009 -= angka2_1009
print("\nAssigmnent Pengurangan")
print("Hasil =", hasil_1009)

# *=
hasil_1009 = angka1_1009
hasil_1009 *= angka2_1009
print("\nAssigmnent Perkalian")
print("Hasil =", hasil_1009)

# /=
if angka2_1009 != 0:
    hasil_1009 = angka1_1009
    hasil_1009 /= angka2_1009
    print("\nAssigmnent Pembagian")
    print("Hasil =", hasil_1009)
    
    hasil_1009 = angka1_1009
    hasil_1009 //= angka2_1009
    print("\nAssigmnent Pembagian Bulat")
    print("Hasil =", hasil_1009)

    hasil_1009 = angka1_1009
    hasil_1009 %= angka2_1009
    print("\nAssigmnent Sisa Bagi")
    print("Hasil =", hasil_1009)
else:
    print("\nPembagian tidak dapat dilakukan")
    print("Angka tidak boleh bernilai nol")
    
````

**Penjelasan :**

Program ini bertugas mendemonstrasikan penggunaan operator penugasan (assignment) sederhana dan compound assignment pada Python, diawali dengan mengambil dua input bilangan bulat dari pengguna serta menampilkan kembali nilai awalnya ke layar.

 Operasi dimulai dengan assignment biasa (=) untuk mengisi nilai variabel hasil dari angka pertama, dilanjutkan dengan compound assignment penambahan (+=), pengurangan (-=), dan perkalian (*=) yang masing-masing secara ringkas menggabungkan operasi aritmatika dengan pembaruan nilai variabel. 
 
 Sebelum menjalankan operasi pembagian, program menggunakan struktur pengondisian if untuk memverifikasi bahwa angka kedua tidak bernilai nol guna mencegah error runtime; jika aman, program mengeksekusi compound assignment pembagian (/=) yang secara otomatis mengubah tipe data variabel hasil menjadi float, pembagian bulat (//=), serta modulo (%=). Jika input angka kedua adalah nol, program akan melompati operasi pembagian dan langsung mencetak pesan bahwa pembagian tidak dapat dilakukan.


### 2.2.5 Program "bitwise_NIM"

````python
print("==============================")
print("     2. OPERATOR BITWISE      ")
print("==============================")

# input angka
angka1_1009 = int(input("input angka 1: "))
angka2_1009 = int(input("input angka 2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_1009, "| biner =", bin(angka1_1009))
print("angka1 =", angka2_1009, "| biner =", bin(angka2_1009))

# Bitwise AND
hasil_1009 = angka1_1009 & angka2_1009
print("\nBitwise AND (&)")
print(angka1_1009, "&", angka2_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))

# Bitwise OR
hasil_1009 = angka1_1009 | angka2_1009
print("\nBitwise OR (|)")
print(angka1_1009, "|", angka2_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))

# Bitwise XOR
hasil_1009 = angka1_1009 ^ angka2_1009
print("\nBitwise XOR (^)")
print(angka1_1009, "^", angka2_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))

# Bitwise NOT
hasil_1009 = ~angka1_1009
print("\nBitwise NOT (~)")
print("~", angka1_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))

 
# Left Shift
jumlah_geser_1009 = int(input("\nMasukkan jumlah pergeseran bit:"))

hasil_1009 = angka1_1009 << jumlah_geser_1009
print("\nBitwise Left Shift (<<)")
print(angka1_1009, "<<", jumlah_geser_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))
 
# Left Shift
hasil_1009 = angka1_1009 >> jumlah_geser_1009
print("\nBitwise Right Shift (>>)")
print(angka1_1009, ">>", jumlah_geser_1009, "=", hasil_1009)
print("\nBiner hasil =", bin(hasil_1009))
print("\nBiner hasil (8 bit) =", format(hasil_1009, "08b"))
 
````

**Penjelasan :**

Program ini mendemonstrasikan manipulasi data pada tingkat biner dengan terlebih dahulu membaca dua input integer dan menampilkan representasi desimal beserta nilai binernya menggunakan fungsi bawaan bin(). 

Program kemudian menjalankan dan menampilkan hasil dari operasi Bitwise AND (&) yang menghasilkan bit 1 jika kedua bit bernilai 1, Bitwise OR (|) yang menghasilkan bit 1 jika salah satu bit bernilai 1, Bitwise XOR (^) yang menghasilkan bit 1 jika kedua bit berbeda, serta Bitwise NOT (~) yang membalikkan seluruh bit dan dihitung secara matematis menggunakan rumus $-(x + 1)$. 

Setiap hasil operasi ditampilkan dalam format nilai desimal dan string biner 8-bit (zero-padded) menggunakan fungsi format(x, "08b"). 

Di bagian akhir, program meminta input jumlah pergeseran bit dari pengguna untuk menjalankan operasi Left Shift (<<) yang menggeser bit ke kiri atau setara dengan perkalian kelipatan $2^n$, serta Right Shift (>>) yang menggeser bit ke kanan atau setara dengan pembagian bulat kelipatan $2^n$.


### 2.2.6 Program "lainnya_NIM"

````python
print("==============================")
print("   1. OPERATOR KEANGGOTAAN    ")
print("==============================")

# input data
input_data_1009 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# input mmenjadi list integer
data_1009 = [int(angka.strip()) for angka in input_data_1009.split(",")]

nilai_dicari_1009 = int(input("masukkan angka yang ingin dicari: "))

# in
hasil_1009 = nilai_dicari_1009 in data_1009
print("\nOperator Keanggotaan IN")
print(f"{nilai_dicari_1009} in {data_1009} = ", hasil_1009)

# operator not in
hasil_1009 = nilai_dicari_1009 not in data_1009
print("\nOperator Keanggotaan NOT IN")
print(f"{nilai_dicari_1009} in {data_1009} = ", hasil_1009)

print("\n=====================================")
print("         2. OPERATOR IDENTITAS       ")
print("=====================================")

# objek1 menggunakan list dari input pengguna
objek1_1009 =data_1009

# objek2 merujuk pada objek yang sama dengan objek1
objek2_1009 = objek1_1009

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_1009 = data_1009.copy()

print("objek1 =", objek1_1009)
print("objek2 =", objek2_1009)
print("objek3 =", objek3_1009)

# Operator is
hasil_1009 = objek1_1009 is objek2_1009
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_1009)

# Operator is not
hasil_1009 = objek1_1009 is not objek3_1009
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_1009)

# membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 == objek3 =", objek1_1009 == objek3_1009)
print("objek1 == objek3 =", objek1_1009 is objek3_1009)
````

**Penjelasan:**

Program ini terbagi menjadi dua bagian utama yang membahas operator keanggotaan dan operator identitas. 

Pada bagian keanggotaan, program menerima sekumpulan angka berbasis string yang dipisahkan oleh koma, kemudian mengonversinya menjadi `list of integers` memanfaatkan list comprehension, dilanjutkan dengan menguji keberadaan sebuah angka acuan di dalam list tersebut menggunakan operator `in` dan `not in`. Pada bagian identitas, program membuat variabel objek2 sebagai alias yang menunjuk ke alamat memori yang sama dengan `objek1`, serta variabel `objek3` yang dibuat menggunakan metode `.copy()` sehingga menghasilkan objek baru dengan isi nilai yang sama tetapi memiliki alamat memori berbeda. Pengujian dengan operator `is` dan `is not` membuktikan keberadaan identitas memori tersebut, di mana bagian penutup program secara tegas memperlihatkan perbedaan antara perbandingan nilai (`==`) yang bernilai `True` karena isi list sama, dengan perbandingan identitas (`is`) yang bernilai `False` karena keduanya bertempat di lokasi memori yang berbeda.

````python
# Program: Sistem Simulasi Transaksi Toko (NIM: 2511531009)

# 1. Input Data Pelanggan & Transaksi
print("=== SISTEM TRANSAKSI TOKO ===\n")

nama_1009 = input("Masukkan Nama Pelanggan : ")
status_input_1009 = input("Masukkan Status Pelanggan (member/nonmember) : ")
total_belanja_1009 = float(input("Masukkan Total Belanja : "))
jumlah_barang_1009 = int(input("Masukkan Jumlah Barang : "))
kode_promo_input_1009 = input("Masukkan Kode Promo : ")

# Break line setelah input terakhir
print()

# Normalisasi teks input
status_1009 = status_input_1009.strip().lower()
kode_promo_1009 = kode_promo_input_1009.strip().upper()

# Daftar promo resmi
promo_tersedia_1009 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]


# 2. Operator Keanggotaan (Membership)
is_promo_valid_1009 = kode_promo_1009 in promo_tersedia_1009
is_promo_invalid_1009 = kode_promo_1009 not in promo_tersedia_1009


# 3. Operator Perbandingan (Comparison)
syarat_belanja_1009 = total_belanja_1009 >= 200000
syarat_barang_1009 = jumlah_barang_1009 >= 3
is_member_1009 = status_1009 == "member"


# 4. Operator Logika (Logical)
dapat_diskon_1009 = is_member_1009 and syarat_belanja_1009
dapat_promo_1009 = (syarat_barang_1009 and is_promo_valid_1009) or dapat_diskon_1009
is_nonmember_1009 = not is_member_1009


# 5. Operator Aritmatika & Penugasan (Assignment)
persen_diskon_1009 = 0.10 if dapat_diskon_1009 else 0.0
besarnya_diskon_1009 = total_belanja_1009 * persen_diskon_1009

total_pembayaran_1009 = total_belanja_1009 - besarnya_diskon_1009

# Augmented assignment (-=) potongan tambahan promo HEMAT10
if is_promo_valid_1009 and kode_promo_1009 == "HEMAT10":
    potongan_promo_1009 = 10000.0
    total_pembayaran_1009 -= potongan_promo_1009
else:
    potongan_promo_1009 = 0.0

rata_rata_harga_1009 = total_pembayaran_1009 / jumlah_barang_1009 if jumlah_barang_1009 > 0 else 0.0
sisa_pembagian_1009 = int(total_pembayaran_1009) % jumlah_barang_1009 if jumlah_barang_1009 > 0 else 0


# 6. Operator Identitas (Identity)
objek_a_1009 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]
objek_b_1009 = promo_tersedia_1009          # Alias (referensi memori sama)
objek_c_1009 = promo_tersedia_1009.copy()   # Duplikat (referensi memori berbeda)

identitas_alias_1009 = objek_b_1009 is promo_tersedia_1009
identitas_copy_1009 = objek_c_1009 is not promo_tersedia_1009
nilai_sama_1009 = objek_c_1009 == promo_tersedia_1009


# 7. Operator Bitwise
# Representasi biner kondisi transaksi
bit_member_1009 = 0b0001 if is_member_1009 else 0b0000
bit_belanja_1009 = 0b0010 if syarat_belanja_1009 else 0b0000
bit_barang_1009 = 0b0100 if syarat_barang_1009 else 0b0000
bit_promo_1009 = 0b1000 if is_promo_valid_1009 else 0b0000

# Bitwise OR (|), AND (&), XOR (^), dan Shift (<< / >>)
kode_status_1009 = bit_member_1009 | bit_belanja_1009 | bit_barang_1009 | bit_promo_1009

cek_member_1009 = kode_status_1009 & 0b0001
cek_promo_1009 = kode_status_1009 & 0b1000

kode_referensi_1009 = 0b1011
beda_status_1009 = kode_status_1009 ^ kode_referensi_1009

shift_left_1009 = kode_status_1009 << 1
shift_right_1009 = kode_status_1009 >> 1


# 8. Output Ringkasan Hasil Program
print("=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan        : {nama_1009}")
print(f"Status Pelanggan      : {status_1009}")
print(f"Total Belanja         : Rp{total_belanja_1009:.0f}")
print(f"Jumlah Barang         : {jumlah_barang_1009}")
print(f"Kode Promo            : {kode_promo_1009}")

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000        : {syarat_belanja_1009}")
print(f"Jumlah Barang >= 3         : {syarat_barang_1009}")
print(f"Status Member              : {is_member_1009}")
print(f"Kode Promo Tersedia        : {is_promo_valid_1009}")
print(f"Mendapatkan Diskon         : {dapat_diskon_1009}")
print(f"Mendapatkan Promo          : {dapat_promo_1009}")

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                     : Rp{besarnya_diskon_1009:.0f}")
print(f"Total Pembayaran           : Rp{total_pembayaran_1009:.0f}")
print(f"Rata-rata Harga Barang     : Rp{rata_rata_harga_1009:.2f}")
print(f"Sisa Bagi Pembayaran % Qty : {sisa_pembagian_1009}")

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Kode Hak Akses             : {format(kode_status_1009, '04b')}")
print(f"Member Access              : {is_member_1009}")
print(f"Promo Access               : {is_promo_valid_1009}")
print(f"Free Shipping Access       : {is_promo_valid_1009 and kode_promo_1009 == 'GRATISONGKIR'}")

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print("0001 | 0010 | 0100 | 1000")
print(f"Kode Biner    : {format(kode_status_1009, '04b')}")
print(f"Kode Desimal  : {kode_status_1009}")

print("\n=== Pemeriksaan Status ===")
print("Cek Member")
print(f"{format(kode_status_1009, '04b')} & 0001")
print(f"Hasil Biner   : {format(cek_member_1009, '04b')}")
print(f"Hasil Desimal : {cek_member_1009}")

print("\nCek Promo")
print(f"{format(kode_status_1009, '04b')} & 1000")
print(f"Hasil Biner   : {format(cek_promo_1009, '04b')}")
print(f"Hasil Desimal : {cek_promo_1009}")

print("\n=== Perbandingan Status ===")
print(f"Kode Transaksi : {format(kode_status_1009, '04b')}")
print(f"Kode Referensi : {format(kode_referensi_1009, '04b')}")
print(f"{format(kode_status_1009, '04b')} ^ {format(kode_referensi_1009, '04b')}")
print(f"Hasil Biner   : {format(beda_status_1009, '04b')}")
print(f"Hasil Desimal : {beda_status_1009}")

print("\n=== Shift ===")
print(f"{format(kode_status_1009, '04b')} << 1")
print(f"Hasil Biner   : {format(shift_left_1009, '05b')}")
print(f"Hasil Desimal : {shift_left_1009}")

print("\n=== SELESAI ===")
````

### Pertanyaan Analisis:

**1. Operator Aritmatika dan Perhitungannya**
Operator aritmatika mengolah komponen harga dan kuantitas dalam transaksi:

* **Pengurangan (`-` ):** Menghitung `total_pembayaran_1009` dengan mengurangkan `total_belanja_1009` dengan `besarnya_diskon_1009`.


* **Perkalian (`*`):** Menghitung nominal diskon melalui `total_belanja_1009 * persen_diskon_1009`.


* **Pembagian (`/`):** Menghitung `rata_rata_harga_1009` per unit barang (`total_pembayaran_1009 / jumlah_barang_1009`).


* **Modulo (`%`):** Menghitung `sisa_pembagian_1009` antara total pembayaran bulat dengan jumlah barang.


* **Pengurangan Penugasan (`-=`):** Memotong langsung total pembayaran sebesar Rp10.000 jika promo `HEMAT10` aktif.


**2. Perbedaan Operator Logika (`and`, `or`, `not`)**

* **`and`:** Bernilai `True` jika **semua** kondisi bernilai `True`. Digunakan pada `dapat_diskon_1009`, di mana pelanggan wajib menjadi member **dan** total belanja $\ge 200.000$.
 
* **`or`:** Bernilai `True` jika **salah satu** kondisi bernilai `True`. Digunakan pada `dapat_promo_1009`, di mana promo didapat jika syarat barang & promo terpenuhi **atau** pelanggan mendapat diskon.
 
* **`not`:** Membalikkan nilai boolean. `is_nonmember_1009` membalikkan status `is_member_1009` (jika member `True`, maka nonmember menjadi `False`).
 
**3. Fungsi Operator `in` dan `not in`**

* **`in`:** Memeriksa apakah input `kode_promo_1009` ada di dalam list `promo_tersedia_1009`. Menghasilkan `True` jika kode valid.


* **`not in`:** Memeriksa apakah input `kode_promo_1009` **tidak ada** di dalam list `promo_tersedia_1009`. Menghasilkan `True` jika kode tidak terdaftar.
 
**4. Perbedaan Operator `is` dan `==`**

* **`==` (Value Equality):** Membandingkan apakah nilai atau isi dari dua objek sama, tanpa peduli lokasi memorinya. Contoh: `objek_c_1009 == promo_tersedia_1009` bernilai `True` karena isi list-nya identik.


* **`is` (Reference Identity):** Membandingkan apakah dua variabel menunjuk ke objek/lokasi memori yang sama (*address*). Contoh: `objek_b_1009 is promo_tersedia_1009` bernilai `True` (alias), sedangkan `objek_c_1009 is not promo_tersedia_1009` bernilai `True` karena `.copy()` membuat objek baru di memori berbeda.

**5. Penggunaan Operator Bitwise untuk Hak Akses**
Hak akses disimpan dalam bentuk *bitmask* 4-bit (`0b0000` hingga `0b1111`):

* Setiap syarat diwakili oleh 1 bit: Member (`0b0001`), Belanja $\ge 200k$ (`0b0010`), Barang $\ge 3$ (`0b0100`), Promo Valid (`0b1000`).

* **Bitwise OR (`|`):** Menggabungkan semua bit kondisi aktif menjadi satu variabel `kode_status_1009`.

* **Bitwise AND (`&`):** Melakukan *masking* untuk mengecek status spesifik, misalnya `kode_status_1009 & 0b0001` untuk mengecek akses member.

**6. Alur Program Transaksi**

1. **Input:** Program menerima 5 data dasar dari pengguna (nama, status, total belanja, jumlah barang, kode promo).


2. **Normalisasi:** Teks diubah formatnya (lowercase/uppercase dan di-strip).


3. **Validasi Kondisi:** Program mengecek status keanggotaan promo (`in`), syarat belanja (`>=`), dan syarat kuantitas barang (`>=`).


4. **Kalkulasi Diskon & Promo:** Menggunakan logika `and`/`or` untuk menentukan persentase diskon (10% atau 0%) dan potongan nominal promo `HEMAT10` (`-= 10000`).


5. **Output Perhitungan:** Menghasilkan `total_pembayaran_1009` akhir beserta rata-rata harga dan sisa bagi.

**7. Identifikasi Kondisi `True` dan `False`**

* **Kondisi Bernilai `True`:** `syarat_belanja_1009` akan bernilai `True` jika pengguna memasukkan Total Belanja sebesar `250000` (karena $250.000 \ge 200.000$).


* **Kondisi Bernilai `False`:** `is_promo_valid_1009` akan bernilai `False` jika pengguna memasukkan Kode Promo `"DISKON50"` (karena `"DISKON50"` tidak ada dalam list `promo_tersedia_1009`).

Berdasarkan praktikum mengenai operator pada bahasa pemrograman Python yang telah dilaksanakan, dapat disimpulkan bahwa:

1. Python menyediakan tujuh kelompok operator utama (aritmatika, assignment, perbandingan, logika, bitwise, keanggotaan, dan identitas) yang masing-masing memiliki sintaks, fungsi, serta tipe data kembalian yang spesifik sesuai kebutuhan pemrosesan data.

2. Operator aritmatika pembagian biasa (/) secara otomatis mengembalikan tipe data float, sedangkan pembagian bulat (//) membulatkan hasil ke bawah. 5. Terdapat perbedaan mendasar antara operator perbandingan nilai (==) dengan operator identitas (is).
